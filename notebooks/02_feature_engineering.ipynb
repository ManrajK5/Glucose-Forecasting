# 02 Feature Engineering

Create lag, rolling, trend, time, and future target columns for model training.

```python
from pathlib import Path
import sys

import pandas as pd

ROOT = Path.cwd().resolve().parent if Path.cwd().name == "notebooks" else Path.cwd().resolve()
sys.path.insert(0, str(ROOT))

from src.features import create_model_features, get_feature_columns

PROCESSED_DIR = ROOT / "data" / "processed"
```

```python
input_path = PROCESSED_DIR / "ohio_all_patients_clean.csv"
clean_df = pd.read_csv(input_path, parse_dates=["timestamp"])
model_df = create_model_features(clean_df, include_targets=True)
print(model_df.shape)
model_df.head()
```

```python
feature_columns = get_feature_columns(model_df)
target_columns = ["target_30min", "target_60min", "target_120min"]
print("Feature columns:", feature_columns)
model_df[feature_columns + target_columns].describe().T
```

```python
output_path = PROCESSED_DIR / "ohio_model_ready.csv"
model_df.to_csv(output_path, index=False)
output_path
```
